```python
import pickle
import pandas as pd

# Specify the path to your pickle file
pkl_file_path = 'kansas_county_daily_weather.pkl'

# Load the pickle file into a dictionary
with open(pkl_file_path, 'rb') as file:
    weather_data = pickle.load(file)

# Combine the list of dataframes into a single dataframe for each county
combined_weather_data = {}

for county, df_list in weather_data.items():
    combined_weather_data[county] = pd.concat(df_list, ignore_index=False)

# Remove specified columns (1st, 3rd, 6th)
for county, df in combined_weather_data.items():
    combined_weather_data[county] = df.drop(df.columns[[0, 2, 5]], axis=1)

# Calculate global min and max values for each column
all_data_combined = pd.concat(combined_weather_data.values())
global_min = all_data_combined.min()
global_max = all_data_combined.max()

# Normalize each column in each dataframe using the global min and max
def normalize(df, global_min, global_max):
    return (df - global_min) / (global_max - global_min)

for county, df in combined_weather_data.items():
    combined_weather_data[county] = normalize(df, global_min, global_max)
```

```python
# Save the normalized dictionary into a new pickle file
normalized_pkl_file_path = 'normalized_weather_data.pkl'
with open(normalized_pkl_file_path, 'wb') as file:
    pickle.dump(combined_weather_data, file)

print(f"Normalized data has been saved to {normalized_pkl_file_path}.")
```
